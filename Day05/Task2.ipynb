```python
super_heroes = {
    "Batman" : {
        "id": 1,
        "aliases": ["Bruce Wayne", "Dark knight"],
        "location": {
            "number" : 1007,
            "street": "Mountain Drive",
            "city": "Gotham"
            }
        },
        "Superman" : {
            "id": 2,
            "aliases": ["Kal-El", "Clark Kent", "The Man of Steel"],
            "location": {
                "number" : 344,
                "street": "Clinton Street",
                "apartment": "3D",
                "city": "Metropolis"
                }
        },
}

print(super_heroes["Superman"]["location"]["city"])
```

Task 2.9

```python
super_heroes["Batman"]["aliases"].append("Caped Crusader")
super_heroes["Wolverine"] = {"id",3}

print(super_heroes)
```

Task 2.10

```python
for hero_names, hero_datas in super_heroes.items():
    print(f"{hero_names}:")

    if "aliases" in hero_datas:
        for alias in hero_datas["aliases"]:
            print(alias)
    else :
        print("No asliases found.")
    print("\n")
```

Task 2.11

```python
input_dict ={
    "dalmatians": 101,
    "pi": 3.14,
    "beast": 666,
    "life": 42,
    "googol": 10^100,
    "jordan": 23,
    "life, the universe and everything": 42,
    "emergency": 911,
    "euler": 2.71828
}

max_key= ""
max_value = 0
for key, value in input_dict.items():
    if value >= max_value:
        max_value = value
        max_key = key

print(max_key)
```
