Task 1.16

```python
first_list = [1,1,1,1,2,2,2,2,2]
second_list = [42,'42', 42.0, 21+21, 42*10/10]
first_duplicated_list = []
second_duplicated_list = []

for item in first_list :
    if item not in first_duplicated_list:
        first_duplicated_list.append(item)

print(first_duplicated_list)

for item in second_list:
    if item not in second_duplicated_list:
        second_duplicated_list.append(item)

print(second_duplicated_list)
```
