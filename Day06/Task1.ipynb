```python
def make_sandwitches(num, veg = False):
    if type(num) == int:
            for i in range(num):
                if veg == True:
                        bread()
                        lettuce()
                        tomato()
                        ham()
                        ham()
                        bread()
                        print()
                else:
                     bread()
                     lettuce()
                     lettuce()
                     tomato()
                     tomato()
                     bread()
                     print()
    else :
        print("I can't do this")

make_sandwitches(2)
make_sandwitches(2,True)
```
