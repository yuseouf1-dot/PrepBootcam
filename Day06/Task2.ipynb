```python
def func_rec(num):
    if num == 1:
        return 1
    return num + func_rec(num - 1)

input_num = int(input())
func_rec(input_num)
```

Task 2.2

```python
def change_string(text):
    result = ""
    for char in text:
        if char.isalpha():    
            result += char.lower()
    return result

def check_string(text):
    if len(text) <= 1:
        return True
    if text[0] != text[-1]:
        return False
    
    return check_string(text[1:-1])

answer = input()
answer_change = change_string(answer)
print(check_string(answer_change))
```

```python
import os

def list_files_recursive(current_path):
    items = os.listdir(current_path)
    for item in items:
        full_path = os.path.join(current_path, item)
        
        print(full_path)
        
        if os.path.isdir(full_path):
            list_files_recursive(full_path)

list_files_recursive('.')
```
