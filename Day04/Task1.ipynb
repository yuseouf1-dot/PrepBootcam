```python
user_integer = int(input("Enter an integer: "))
result = ""
if user_integer == 42:
    result += "a"
if user_integer <= 21:
    result = "b"
if user_integer % 2 == 0:
    result += "c"
if user_integer / 2  < 21:
    result += "d"
if user_integer % 2 == 1 and user_integer >= 45:
    result += "e"
if result == "":
    result = "f"
print(result)
```
